# Exercicos condicionais

Faça um algoritmo para ler dois números e imprimir o maior, o menor ou então dizer se são iguais.

Algoritmo:

1. Leia o primeiro número e salve na variável num1
2. Leia o segundo número e salve na variável num2
3. Descobrir o maior
4. Mostrar o maior
5. Descobrir o menor
6. Mostrar o menor
7. Descobir se são iguais
8. Mostrar se são iguais

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
if num1 > num2:
    print(f"O maior número é {num1}")
    print(f"O menor número é {num2}")
elif num1 < num2:
    print(f"O maior número é {num2}")
    print(f"O menor número é {num1}")
else:
    print("Os números são iguais")

```

4. Faça um algoritmo para ler um número inteiro e dizer se o número lido é par ou impar.

Algoritmo:

1. Leia o primeiro número e salve na variável num
2. Descobrir se é par ou impar: dividindo por 2 e se tiver resto o número é ímpar.
3. Mostrar o resultado.



```python
num = int(input("Digite um número: "))
if num % 2 == 0:
    print(f"O número {num} é par")
else:
    print(f"O número {num} é ímpar")
```
